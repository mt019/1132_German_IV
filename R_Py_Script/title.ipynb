```python
import os
import re

def replace_headings_in_markdown(folder_path):
    pattern = re.compile(r'^(#{1,6}) 第\s*(\d+)\s*句：(.+)$', re.MULTILINE)

    all_files = []
    modified_files = []

    for root, dirs, files in os.walk(folder_path):
        for file in files:
            if file.endswith('.md'):
                file_path = os.path.join(root, file)
                all_files.append(file_path)

                with open(file_path, 'r', encoding='utf-8') as f:
                    content = f.read()

                new_content = pattern.sub(
                    lambda m: f"{m.group(1)} S.{int(m.group(2)):02d}\n\n{m.group(3).strip()}",
                    content
                )

                if new_content != content:
                    modified_files.append(file_path)
                    with open(file_path, 'w', encoding='utf-8') as f:
                        f.write(new_content)

    print("\n📁 所有遍歷過的 Markdown 檔案：")
    for file in all_files:
        print(" -", file)

    print("\n✅ 有修改內容的檔案：")
    if modified_files:
        for file in modified_files:
            print(" *", file)
    else:
        print("（沒有任何檔案被修改）")

# ✅ 使用方式（請替換成你實際的資料夾路徑）
replace_headings_in_markdown('/Users/iw/Documents/NTU/1132/1132_German_IV/01_Notes/KB_60')
```

```python
import os
import re

def replace_translation_headings(folder_path):
    # 支援 ### 原文 / ### 翻譯（允許前後有空格）
    pattern = re.compile(r'^(#{1,6})\s*(原文|翻譯)\s*$', re.MULTILINE)

    all_files = []
    modified_files = []

    for root, dirs, files in os.walk(folder_path):
        for file in files:
            if file.endswith('.md'):
                file_path = os.path.join(root, file)
                all_files.append(file_path)

                with open(file_path, 'r', encoding='utf-8') as f:
                    content = f.read()

                # 替換成粗體格式
                new_content = pattern.sub(lambda m: f"**{m.group(2)}**\n", content)

                if new_content != content:
                    modified_files.append(file_path)
                    with open(file_path, 'w', encoding='utf-8') as f:
                        f.write(new_content)

    # 顯示所有處理過的檔案
    print("\n📁 遍歷過的 Markdown 檔案：")
    for f in all_files:
        print(" -", f)

    # 顯示有實際修改的檔案
    print("\n✅ 有修改內容的檔案：")
    if modified_files:
        for f in modified_files:
            print(" *", f)
    else:
        print("（沒有任何檔案被修改）")

# ✅ 請修改成你實際的資料夾路徑
replace_translation_headings('/Users/iw/Documents/NTU/1132/1132_German_IV')
```
