```python
def custom_sort_key(word):
    return (word[0].islower(), word.lower())

def sort_and_format_vocab_file(md_path, output_path=None):
    output_path = output_path or md_path

    with open(md_path, 'r', encoding='utf-8') as f:
        lines = f.readlines()

    # 去掉空行與前後空白，只保留有內容的詞
    words = [line.strip() for line in lines if line.strip()]

    # 使用自訂排序規則：小寫詞先，大寫詞後；各自再按字母序
    words_sorted = sorted(words, key=custom_sort_key)

    # 插入每個詞之間的空行
    formatted = []
    for word in words_sorted:
        formatted.append(word)
        formatted.append("")  # 加入空行

    # 寫入結果
    with open(output_path, 'w', encoding='utf-8') as f:
        f.write('\n'.join(formatted))

    print(f"已排序並格式化詞彙，儲存至：{output_path}")

    
sort_and_format_vocab_file(
    md_path="/Users/iw/Documents/NTU/1132/1132_German_IV/01_Notes/kb_vocab_frequency.md",
    output_path="/Users/iw/Documents/NTU/1132/1132_German_IV/01_Notes/kb_vocab_frequency.md"  # 可省略表示覆蓋原檔
)
```
