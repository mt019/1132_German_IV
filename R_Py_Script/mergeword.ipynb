```python
import os
from docx import Document

def merge_docx_files(folder_path, output_path, include_subfolders=True):
    merged_document = Document()

    def append_doc(source_doc):
        for element in source_doc.element.body:
            merged_document.element.body.append(element)

    for root, dirs, files in os.walk(folder_path):
        for filename in sorted(files):
            if filename.endswith(".docx"):
                file_path = os.path.join(root, filename)
                print(f"Merging: {file_path}")
                doc = Document(file_path)
                append_doc(doc)
                merged_document.add_page_break()
        if not include_subfolders:
            break  # only process the top-level folder

    merged_document.save(output_path)
    print(f"✅ Merged document saved to: {output_path}")

# === USAGE ===
# Replace the folder below with your own folder path
source_folder = "/Users/iw/Documents/NTU/1132/1132_German_IV/course_files_export"     # ← path to folder with .docx files
output_file = "merged_word_output.docx"              # ← desired output file name

merge_docx_files(source_folder, output_file)
```
