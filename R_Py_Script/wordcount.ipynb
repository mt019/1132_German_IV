```python
import re
import csv
from collections import Counter

def count_words_to_csv(file_path, output_csv_path, top_n=None):
    with open(file_path, 'r', encoding='utf-8') as f:
        text = f.read()

    # 清除 Markdown 標記與標點，只保留字母、德文字母與 -
    text = re.sub(r'[^\wäöüÄÖÜß\-]+', ' ', text)
    words = text.split()

    # 統計單字
    word_freq = Counter(words)
    sorted_words = word_freq.most_common(top_n)

    # 寫入 CSV
    with open(output_csv_path, 'w', encoding='utf-8', newline='') as csvfile:
        writer = csv.writer(csvfile)
        writer.writerow(['Word', 'Frequency'])
        writer.writerows(sorted_words)

    print(f"成功儲存至：{output_csv_path}")

# 用法範例：請根據實際路徑修改檔案位置
input_path = "/Users/iw/Documents/NTU/1132/1132_German_IV/01_Notes/_KB_Source-linebreak.md"
output_path = "/Users/iw/Documents/NTU/1132/1132_German_IV/01_Notes/kb_vocab_frequency.csv"

count_words_to_csv(input_path, output_path, top_n=None)  # 設定 top_n=100 可限制輸出前百大詞
```
